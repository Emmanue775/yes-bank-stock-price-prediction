## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
# Convert Date column into datetime format
data['Date'] = pd.to_datetime(data['Date'], format='%b-%y')
# Check updated data types
data.dtypes
```

```python
# Sort dataset by Date
data = data.sort_values(by='Date')
# Reset index
data = data.reset_index(drop=True)
# Display first five records
data.head()
```

```python
# Check duplicate dates
data['Date'].duplicated().sum()
```

```python

# Check whether High is greater than or equal to Open and Close
print("Invalid High values:",
      ((data['High'] < data['Open']) |
       (data['High'] < data['Close'])).sum())

# Check whether Low is less than or equal to Open and Close
print("Invalid Low values:",
      ((data['Low'] > data['Open']) |
       (data['Low'] > data['Close'])).sum())
```

```python
# Check whether High is greater than or equal to Low
(data['High'] < data['Low']).sum()
```

### What all manipulations have you done and insights you found?

### Data Wrangling – Manipulations Performed

1. **Date Conversion:** Converted the Date column from object data type to datetime64 format to support time-based analysis.

2. **Sorting:** Arranged the dataset in chronological order based on the Date column.

3. **Index Resetting:** Reset the DataFrame index after sorting to maintain a sequential index.

4. **Missing Values:** Checked all columns and found zero missing values.

5. **Duplicate Records:** Checked for completely duplicated rows and found zero duplicates.

6. **Duplicate Dates:** Verified that there are no duplicate dates in the dataset.

7. **OHLC Validation:** Checked the relationships between Open, High, Low and Close prices to identify invalid stock-price records. No inconsistencies were found in the tested conditions.

### Key Insights

- The dataset contains 185 records and 5 columns.
- All records contain complete data.
- No duplicate records or duplicate dates were identified.
- The High price is greater than or equal to Open, Close and Low for all records.
- The Low price is less than or equal to Open and Close for all records.
- No records needed to be removed during the initial data-wrangling process.

**Conclusion:** The dataset passed the initial data-quality and OHLC validation checks. It is ready for further Exploratory Data Analysis (EDA).
